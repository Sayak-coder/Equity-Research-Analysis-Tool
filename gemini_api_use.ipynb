```python
from langchain_classic.chains import LLMChain
from langchain_core.prompts import PromptTemplate

prompt_template_1 = PromptTemplate(
    input_variables = ['country'],
    template = "who is the presedent of {country} ?"
)

name_chain = LLMChain(
    llm = llm,
    prompt = prompt_template_1,
    output_key = "president_name"
)

prompt_template_2 = PromptTemplate(
    input_variables=['president_name'],
    template = "when was {president_name} born ?"
)

age_chain = LLMChain(
    llm = llm,
    prompt = prompt_template_2,
    output_key = "age"
)
```

```python
from langchain_classic.chains import SimpleSequentialChain

chain = SimpleSequentialChain(
    chains = [name_chain, age_chain]
)

response = chain.invoke({"India"})

print(response)
```

```python
from google.colab import userdata
userdata.get('groq')
```

```python
from langchain_groq import ChatGroq

model = 'openai/gpt-oss-20b'
llm = ChatGroq(model = model, groq_api_key = userdata.get('groq'),temperature=0.6)
```

```python
from langchain_core.prompts import PromptTemplate

prompt_template_name = PromptTemplate(
    input_variables=['country'],
    template = "What is the name of the president of {country}"
)

prompt_template_age = PromptTemplate(
    input_variables = ['president_name'],
    template = "What is the age of {president_name}"
)
```

```python
chain = prompt_template_name | llm

response = chain.invoke({'country': 'India'})
print(response.content)
```

```python
print(type(response))
```

```python
from langchain_classic.chains import LLMChain

name_chain = LLMChain(
    llm = llm,
    prompt = prompt_template_name,
    output_key = 'president_name'
)

age_chain = LLMChain(
    llm = llm,
    prompt = prompt_template_age,
    output_key = 'age'
)
```

```python
from langchain_classic.chains import SimpleSequentialChain

name_chain = SimpleSequentialChain(
    chains = [name_chain, age_chain]
)

response = name_chain.invoke("India")
print(response)
```

```python
from langchain_classic.chains import SequentialChain

sequential_chain = SequentialChain(
    chains = [name_chain, age_chain],
    input_variables = ['country'],
    output_variables = ['president_name', 'age']
)

response = sequential_chain.invoke("India")
print(response)
```

```python
from langchain_core.prompts import PromptTemplate
from langchain_classic.agents import initialize_agent, AgentType
from langchain_community.agent_toolkits.load_tools import load_tools
```

```python
from langgraph.prebuilt import create_react_agent

tools = load_tools(["serpapi","wikipedia","llm-math"], llm = llm)

agent = create_react_agent(llm, tools)

result = agent.invoke(
    {"messages": [("user", "What was the GDP of India in 2024?")]}
)
print(result["messages"][-1].content)
```

#Memory

```python
from langchain_classic.chains import LLMChain

chain = LLMChain(llm = llm, prompt = prompt_template_name, output_key = 'president_name')
```

```python
name = chain.invoke("India")
print(name)
```

```python
from langchain_classic.memory import ConversationBufferWindowMemory
from langchain_classic.chains import ConversationChain

memory = ConversationBufferWindowMemory(k=1)

chain = ConversationChain(
    llm = llm,
    memory = memory
)
```

```python
answer = chain.invoke("Who won 2023 cricket men worldcup?")
print(answer)
```

```python
answer = chain.invoke("who was the captain ?")
print(answer)
```
